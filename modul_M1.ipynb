# MODUL 1 — LAYERED CONSOLIDATION SETTLEMENT

# M1 — Layered Consolidation Settlement

## 1. Objective and project context

M1 calculates **primary one-dimensional consolidation settlement** for a compressible soil deposit represented by horizontal sublayers. The module is organized into three calculation stages:

1. `calculate_initial_stress()` — calculates initial total stress, hydrostatic pore-water pressure, and initial effective stress at each layer center.
2. `calculate_stress_increment()` — calculates the effective-stress increment associated with the C-2 vacuum-preloading mechanism from the change in pore-water pressure, with an optional total-stress increment where a separately established fill/surcharge load is present.
3. `calculate_layered_settlement()` — calculates primary consolidation settlement for each layer according to its stress history (NC or OC), then sums the layer settlements.

The module is contextualized to **C-2: vacuum preloading at the Palembang–Indralaya toll-road project**. It is not a complete reproduction of the project design because the presentation does not provide all soil-compressibility parameters or a numerical pore-pressure profile at every layer center.

## 2. Governing effective-stress relationship

Terzaghi's effective-stress relationship is used:

\[
\sigma'_v = \sigma_v-u
\]

For a change in loading and pore-water pressure:

\[
\boxed{\Delta \sigma'_v = \Delta \sigma_v-\Delta u}
\]

where:

- \(\Delta \sigma_v\) = change in total vertical stress (kPa);
- \(\Delta u=u_f-u_0\) = change in pore-water pressure (kPa);
- \(\Delta \sigma'_v\) = change in effective vertical stress (kPa).

For the vacuum contribution, the presentation describes the generation of negative pore-water pressure. If total stress is unchanged:

\[
\boxed{\Delta \sigma'_{v,\mathrm{vac}}=-\Delta u_{\mathrm{vac}}}
\]

A negative \(\Delta u\) therefore produces a positive effective-stress increment.

If fill is placed above the membrane during vacuum operation, its total-stress contribution may also be included:

\[
\Delta \sigma'_v
=
\Delta \sigma_{v,\mathrm{fill}}-\Delta u_{\mathrm{vac}}
\]

The fill contribution must be calculated independently from actual fill geometry and unit weight. It must not be inferred from the vacuum pressure.

**C-2 source:** Irsyam et al., *Pengalaman Perbaikan Tanah untuk Beberapa Proyek Infrastruktur di Indonesia*, Seminar Bina Marga (2018), PDF pp. 88 and 93. The slides describe negative pore-water pressure under vacuum and list filling above the membrane during vacuum and monitoring.

## 3. Layered consolidation settlement

The Overconsolidation Ratio is:

\[
OCR_i=\frac{\sigma'_{p,i}}{\sigma'_{v0,i}}
\]

where \(\sigma'_{p,i}\) is the preconsolidation pressure and \(\sigma'_{v0,i}\) is the initial effective vertical stress.

### 3.1 Normally consolidated soil (NC)

For \(OCR=1\):

\[
S_{c,i}=
\frac{C_{c,i}H_i}{1+e_{0,i}}
\log_{10}\left(\frac{\sigma'_{vf,i}}{\sigma'_{v0,i}}\right)
\]

### 3.2 Overconsolidated soil (OC): recompression only

When \(\sigma'_{vf,i}\leq \sigma'_{p,i}\):

\[
S_{c,i}=
\frac{C_{r,i}H_i}{1+e_{0,i}}
\log_{10}\left(\frac{\sigma'_{vf,i}}{\sigma'_{v0,i}}\right)
\]

This expression follows from the recompression slope \(C_r\) of the \(e-\log \sigma'\) relationship.

### 3.3 Overconsolidated soil (OC): recompression and virgin compression

When \(\sigma'_{vf,i}>\sigma'_{p,i}\):

\[
S_{c,i}=
\frac{H_i}{1+e_{0,i}}
\left[
C_{r,i}\log_{10}\left(\frac{\sigma'_{p,i}}{\sigma'_{v0,i}}\right)
+
C_{c,i}\log_{10}\left(\frac{\sigma'_{vf,i}}{\sigma'_{p,i}}\right)
\right]
\]

The final effective stress is:

\[
\sigma'_{vf,i}=\sigma'_{v0,i}+\Delta\sigma'_{v,i}
\]

Total primary consolidation settlement:

\[
S_c=\sum_{i=1}^{n}S_{c,i}
\]

## 4. Initial stress calculation

At the center of each layer:

\[
\sigma'_{v0}=\sigma_{v0}-u_0
\]

For hydrostatic groundwater conditions:

\[
u_0=\gamma_w\max(z-z_{GWT},0)
\]

Total stress is integrated from the ground surface to the evaluation depth, using moist unit weight above the groundwater table and saturated unit weight below it.

## 6. Scope, assumptions, and exclusions

1. Soil is represented by horizontal sublayers; parameters are constant within each layer.
2. Initial stresses are evaluated at layer centers.
3. Initial pore-water pressure is hydrostatic; capillary suction is excluded.
4. Vacuum stress increment is calculated from the **measured or otherwise justified pore-pressure change at the relevant depth**. Surface pump vacuum is not assumed to be fully transferred to every layer.
5. Optional fill/surcharge contribution is supplied as a separately calculated total-stress increment for each layer.
6. The settlement engine accepts effective-stress increments and calculates primary consolidation only.
7. Immediate settlement, secondary compression, lateral deformation, smear effects, well resistance, and time-rate consolidation are outside M1.
8. OCR below 1 is rejected by the current NC/OC settlement model.
9. The supplied C-2 presentation does not contain enough numerical soil-property and pore-pressure data to produce a fully project-specific M1 settlement prediction. Actual C-2 calculation requires supplementary investigation/laboratory/monitoring data.

## Main Functions

The functions are defined separately. All function definitions appear before the testing cells.

```python
import math
import pandas as pd


def calculate_layered_settlement(
    layer_thickness,
    ocr,
    cc,
    cr,
    e0,
    sigma_v0_eff,
    delta_sigma_eff
):
    """
    Calculate primary one-dimensional consolidation settlement by layer.

    Inputs
    ------
    layer_thickness : m
    ocr, cc, cr, e0 : dimensionless
    sigma_v0_eff, delta_sigma_eff : kPa (effective vertical stress)

    Returns
    -------
    dict
        Layer classification, preconsolidation/final stress, settlement
        per layer, and total primary settlement (m and mm).

    Governing equations
    -------------------
    NC:
        Sc = H*Cc/(1+e0) * log10(sigma_vf/sigma_v0)

    OC recompression only:
        Sc = H*Cr/(1+e0) * log10(sigma_vf/sigma_v0)

    OC recompression + virgin compression:
        Sc = H/(1+e0) * [
            Cr*log10(sigma_p/sigma_v0)
            + Cc*log10(sigma_vf/sigma_p)
        ]

    Traceability
    ------------
    PennDOT (2025), Publication 293, Chapter 7, Sections 7.5.2.1
    and 7.5.2.2, pp. 7-21 to 7-22.
    Kimmerling (2002), FHWA-SA-02-054, Chapter 5, Section 5.3.3,
    Equations 5-20 and 5-22, pp. 82-84.

    Scope
    -----
    Primary 1D consolidation only. Immediate and secondary settlement
    are excluded. OCR < 1 is outside the supported NC/OC model.
    """

    inputs = {
        "layer_thickness": layer_thickness,
        "ocr": ocr,
        "cc": cc,
        "cr": cr,
        "e0": e0,
        "sigma_v0_eff": sigma_v0_eff,
        "delta_sigma_eff": delta_sigma_eff
    }

    for name, values in inputs.items():
        if not hasattr(values, "__len__"):
            raise TypeError(f"{name} must be a sequence of numeric values.")

    lengths = {name: len(values) for name, values in inputs.items()}
    if len(set(lengths.values())) != 1:
        raise ValueError(f"All input arrays must have equal length. Received: {lengths}")

    n = len(layer_thickness)
    if n == 0:
        raise ValueError("At least one soil layer is required.")

    for name, values in inputs.items():
        for i, value in enumerate(values):
            if not isinstance(value, (int, float)) or not math.isfinite(value):
                raise ValueError(f"{name}[{i}] must be a finite numeric value.")

    for i in range(n):
        if layer_thickness[i] <= 0:
            raise ValueError(f"Layer {i+1}: thickness must be > 0.")
        if ocr[i] < 1:
            raise ValueError(f"Layer {i+1}: OCR < 1 is outside the supported NC/OC model.")
        if cc[i] <= 0:
            raise ValueError(f"Layer {i+1}: Cc must be > 0.")
        if cr[i] < 0:
            raise ValueError(f"Layer {i+1}: Cr must be >= 0.")
        if e0[i] <= -1:
            raise ValueError(f"Layer {i+1}: e0 must be > -1.")
        if sigma_v0_eff[i] <= 0:
            raise ValueError(f"Layer {i+1}: initial effective stress must be > 0.")
        if delta_sigma_eff[i] < 0:
            raise ValueError(f"Layer {i+1}: effective stress increment must be >= 0.")

    layer_results = []

    for i in range(n):
        H = float(layer_thickness[i])
        OCR = float(ocr[i])
        Cc = float(cc[i])
        Cr = float(cr[i])
        e0_i = float(e0[i])
        sigma_0 = float(sigma_v0_eff[i])
        delta_sigma = float(delta_sigma_eff[i])

        sigma_p = OCR * sigma_0
        sigma_f = sigma_0 + delta_sigma

        if math.isclose(OCR, 1.0, rel_tol=1e-9, abs_tol=1e-12):
            classification = "NC"
            mechanism = "Virgin compression"
            settlement = (
                H * Cc / (1.0 + e0_i) * math.log10(sigma_f / sigma_0)
                if sigma_f > sigma_0 else 0.0
            )

        elif sigma_f <= sigma_p or math.isclose(sigma_f, sigma_p, rel_tol=1e-12, abs_tol=1e-12):
            classification = "OC"
            mechanism = "Recompression only"
            settlement = (
                H * Cr / (1.0 + e0_i) * math.log10(sigma_f / sigma_0)
                if sigma_f > sigma_0 else 0.0
            )

        else:
            classification = "OC"
            mechanism = "Recompression + Virgin compression"
            settlement = (
                H / (1.0 + e0_i)
                * (
                    Cr * math.log10(sigma_p / sigma_0)
                    + Cc * math.log10(sigma_f / sigma_p)
                )
            )

        layer_results.append({
            "layer": i + 1,
            "thickness_m": H,
            "ocr": OCR,
            "classification": classification,
            "mechanism": mechanism,
            "sigma_v0_eff_kpa": sigma_0,
            "sigma_p_kpa": sigma_p,
            "delta_sigma_eff_kpa": delta_sigma,
            "sigma_vf_eff_kpa": sigma_f,
            "settlement_m": settlement,
            "settlement_mm": settlement * 1000.0
        })

    total_settlement_m = sum(row["settlement_m"] for row in layer_results)

    return {
        "number_of_layers": n,
        "layer_results": layer_results,
        "total_settlement_m": total_settlement_m,
        "total_settlement_mm": total_settlement_m * 1000.0
    }
```

### Function 2 — Initial Effective Stress

```python
def calculate_initial_stress(
    layer_thickness,
    gamma_moist,
    gamma_sat,
    groundwater_depth,
    gamma_w=9.81
):
    """
    Calculate initial total stress, hydrostatic pore pressure, and
    effective vertical stress at each layer center.

    Inputs
    ------
    layer_thickness : m
    gamma_moist, gamma_sat, gamma_w : kN/m3
    groundwater_depth : m below ground surface

    Returns
    -------
    dict
        Layer boundaries/centers and sigma_v0, u0, sigma'_v0 (kPa).

    Governing equations
    -------------------
    sigma_v0 = vertical integration of moist/saturated unit weight
    u0 = gamma_w * max(z_center - groundwater_depth, 0)
    sigma'_v0 = sigma_v0 - u0

    Assumptions
    -----------
    Hydrostatic groundwater; no capillary suction; stresses evaluated
    at layer centers. Soil unit weights are constant within each layer.

    Traceability
    ------------
    Effective-stress and settlement framework: PennDOT (2025),
    Publication 293, Chapter 7, Sections 7.3 and 7.5.2.
    """

    inputs = {
        "layer_thickness": layer_thickness,
        "gamma_moist": gamma_moist,
        "gamma_sat": gamma_sat
    }

    lengths = {name: len(values) for name, values in inputs.items()}
    if len(set(lengths.values())) != 1:
        raise ValueError(f"All soil profile arrays must have equal length: {lengths}")

    n = len(layer_thickness)
    if n == 0:
        raise ValueError("At least one soil layer is required.")

    if not isinstance(groundwater_depth, (int, float)) or not math.isfinite(groundwater_depth) or groundwater_depth < 0:
        raise ValueError("Groundwater depth must be finite and >= 0.")
    if not isinstance(gamma_w, (int, float)) or not math.isfinite(gamma_w) or gamma_w <= 0:
        raise ValueError("Water unit weight must be finite and > 0.")

    for name, values in inputs.items():
        for i, value in enumerate(values):
            if not isinstance(value, (int, float)) or not math.isfinite(value):
                raise ValueError(f"{name}[{i}] must be a finite numeric value.")

    for i in range(n):
        if layer_thickness[i] <= 0:
            raise ValueError(f"Layer {i+1}: thickness must be > 0.")
        if gamma_moist[i] <= 0 or gamma_sat[i] <= 0:
            raise ValueError(f"Layer {i+1}: unit weights must be > 0.")
        if gamma_sat[i] < gamma_moist[i]:
            raise ValueError(f"Layer {i+1}: gamma_sat must be >= gamma_moist.")

    layer_results = []
    layer_top = 0.0
    boundaries = []
    for H in layer_thickness:
        boundaries.append((layer_top, layer_top + H))
        layer_top += H

    for i, (top, bottom) in enumerate(boundaries):
        center = (top + bottom) / 2.0
        sigma_v0 = 0.0

        for j, (segment_top, segment_bottom) in enumerate(boundaries[:i + 1]):
            segment_end = min(segment_bottom, center)
            if segment_end <= segment_top:
                continue

            above_gwt = max(
                0.0,
                min(segment_end, groundwater_depth) - segment_top
            )
            below_gwt = max(
                0.0,
                segment_end - max(segment_top, groundwater_depth)
            )

            sigma_v0 += (
                gamma_moist[j] * above_gwt
                + gamma_sat[j] * below_gwt
            )

        u0 = gamma_w * max(center - groundwater_depth, 0.0)
        sigma_v0_eff = sigma_v0 - u0

        layer_results.append({
            "layer": i + 1,
            "depth_top_m": top,
            "depth_bottom_m": bottom,
            "depth_center_m": center,
            "sigma_v0_kpa": sigma_v0,
            "u0_kpa": u0,
            "sigma_v0_eff_kpa": sigma_v0_eff
        })

    return {
        "number_of_layers": n,
        "groundwater_depth_m": float(groundwater_depth),
        "layer_results": layer_results
    }
```

### Function 3 — Vacuum Stress Increment (C-2)

```python
def calculate_stress_increment(
    initial_pore_pressure,
    final_pore_pressure,
    delta_sigma_total=None
):
    """
    Calculate layer-wise effective vertical stress increment for C-2
    vacuum preloading using measured/justified pore-pressure changes.

    Parameters
    ----------
    initial_pore_pressure : sequence[float]
        Initial pore-water pressure at each layer center (kPa).
    final_pore_pressure : sequence[float]
        Pore-water pressure at the selected time during/after vacuum
        application, at the same layer centers (kPa).
    delta_sigma_total : sequence[float], optional
        Independently calculated total vertical stress increment at
        each layer center (kPa), e.g. from fill placed above membrane.
        Defaults to zero for vacuum-only contribution.

    Returns
    -------
    dict
        Per-layer delta_u, delta_sigma_total, delta_sigma_eff and
        final effective-stress increment.

    Governing relationship
    ----------------------
        delta_u = u_final - u_initial
        delta_sigma_eff = delta_sigma_total - delta_u

    For vacuum-only loading with unchanged total stress:
        delta_sigma_eff = -delta_u

    Notes
    -----
    - Pore pressures must be signed consistently as gauge/pressure
      values using the same datum and convention.
    - Vacuum-induced pore-pressure change is expected to be <= 0.
    - Surface pump vacuum is NOT assumed to transfer fully to depth.
    - Any fill/surcharge contribution must be calculated separately.
    """

    if delta_sigma_total is None:
        delta_sigma_total = [0.0] * len(initial_pore_pressure)

    inputs = {
        "initial_pore_pressure": initial_pore_pressure,
        "final_pore_pressure": final_pore_pressure,
        "delta_sigma_total": delta_sigma_total
    }

    for name, values in inputs.items():
        if not hasattr(values, "__len__"):
            raise TypeError(f"{name} must be a sequence of numeric values.")

    lengths = {name: len(values) for name, values in inputs.items()}
    if len(set(lengths.values())) != 1:
        raise ValueError(f"All pore-pressure/stress arrays must have equal length: {lengths}")

    n = len(initial_pore_pressure)
    if n == 0:
        raise ValueError("At least one layer-center pressure value is required.")

    layer_results = []

    for i in range(n):
        u0 = initial_pore_pressure[i]
        uf = final_pore_pressure[i]
        delta_sigma = delta_sigma_total[i]

        for name, value in (
            ("initial_pore_pressure", u0),
            ("final_pore_pressure", uf),
            ("delta_sigma_total", delta_sigma)
        ):
            if not isinstance(value, (int, float)) or not math.isfinite(value):
                raise ValueError(f"{name}[{i}] must be a finite numeric value.")

        delta_u = uf - u0

        if delta_u > 1e-9:
            raise ValueError(
                f"Layer {i+1}: pore pressure increased (delta_u={delta_u:.6f} kPa). "
                "This is not a vacuum-induced pore-pressure reduction."
            )

        if delta_sigma < 0:
            raise ValueError(
                f"Layer {i+1}: delta_sigma_total must be >= 0 for this loading model."
            )

        delta_sigma_eff = delta_sigma - delta_u

        if delta_sigma_eff < -1e-9:
            raise ValueError(
                f"Layer {i+1}: calculated effective stress increment is negative."
            )

        delta_sigma_eff = max(delta_sigma_eff, 0.0)

        layer_results.append({
            "layer": i + 1,
            "initial_pore_pressure_kpa": float(u0),
            "final_pore_pressure_kpa": float(uf),
            "delta_u_kpa": delta_u,
            "delta_sigma_total_kpa": float(delta_sigma),
            "delta_sigma_eff_kpa": delta_sigma_eff
        })

    return {
        "number_of_layers": n,
        "loading_model": "Vacuum preloading with optional independently calculated fill load",
        "layer_results": layer_results
    }
```

### Integrated M1 Workflow

```python
def calculate_m1_vacuum_settlement(
    layer_thickness,
    gamma_moist,
    gamma_sat,
    groundwater_depth,
    ocr,
    cc,
    cr,
    e0,
    final_pore_pressure,
    delta_sigma_total=None,
    gamma_w=9.81
):
    """
    Run the complete M1 sequence for the C-2 vacuum-preloading model.

    Calculation sequence
    --------------------
    1. Initial total/effective stress at each layer center.
    2. Effective-stress increment from pore-pressure reduction and,
       optionally, separately calculated fill/surcharge stress.
    3. Layered primary consolidation settlement.

    final_pore_pressure must be supplied at each layer center for the
    selected analysis time. It must use the same pressure datum and
    sign convention as the hydrostatic initial pore pressure.

    This wrapper does not infer depth transfer from surface pump vacuum.
    """

    initial_stress = calculate_initial_stress(
        layer_thickness=layer_thickness,
        gamma_moist=gamma_moist,
        gamma_sat=gamma_sat,
        groundwater_depth=groundwater_depth,
        gamma_w=gamma_w
    )

    initial_u = [
        row["u0_kpa"] for row in initial_stress["layer_results"]
    ]

    stress_increment = calculate_stress_increment(
        initial_pore_pressure=initial_u,
        final_pore_pressure=final_pore_pressure,
        delta_sigma_total=delta_sigma_total
    )

    sigma_v0_eff = [
        row["sigma_v0_eff_kpa"] for row in initial_stress["layer_results"]
    ]

    delta_sigma_eff = [
        row["delta_sigma_eff_kpa"] for row in stress_increment["layer_results"]
    ]

    settlement = calculate_layered_settlement(
        layer_thickness=layer_thickness,
        ocr=ocr,
        cc=cc,
        cr=cr,
        e0=e0,
        sigma_v0_eff=sigma_v0_eff,
        delta_sigma_eff=delta_sigma_eff
    )

    return {
        "initial_stress": initial_stress,
        "stress_increment": stress_increment,
        "settlement": settlement,
        "layer_results": [
            {
                "layer": i + 1,
                "depth_center_m": initial_stress["layer_results"][i]["depth_center_m"],
                "sigma_v0_eff_kpa": sigma_v0_eff[i],
                "delta_sigma_eff_kpa": delta_sigma_eff[i],
                "sigma_vf_eff_kpa": settlement["layer_results"][i]["sigma_vf_eff_kpa"],
                "settlement_mm": settlement["layer_results"][i]["settlement_mm"]
            }
            for i in range(len(layer_thickness))
        ],
        "total_settlement_mm": settlement["total_settlement_mm"]
    }
```

## Function Testing and Integration Testing

The tests below verify individual functions against independently calculated analytical values. Synthetic test inputs are used only to verify implementation; they are **not** represented as measured C-2 project data.


```python
# ============================================================
# M1 - LAYERED SETTLEMENT FUNCTION TESTING
# ============================================================

settlement_test_results = []


def record_settlement_test(name, description, actual, expected, tolerance=1e-10):
    difference = abs(actual - expected)
    relative_error = difference / abs(expected) * 100 if expected != 0 else 0.0
    passed = math.isclose(actual, expected, rel_tol=tolerance, abs_tol=tolerance)

    settlement_test_results.append({
        "Test Case": name,
        "Description": description,
        "Expected (mm)": expected * 1000.0,
        "Actual (mm)": actual * 1000.0,
        "Difference (mm)": difference * 1000.0,
        "Relative Error (%)": relative_error,
        "Status": "PASS" if passed else "FAIL"
    })

    assert passed, (
        f"{name} FAILED: expected {expected:.12f} m, "
        f"actual {actual:.12f} m."
    )


# TEST 1 — Normally consolidated soil (NC)
result_nc = calculate_layered_settlement(
    layer_thickness=[2.0],
    ocr=[1.0],
    cc=[0.30],
    cr=[0.05],
    e0=[0.80],
    sigma_v0_eff=[100.0],
    delta_sigma_eff=[100.0]
)
expected_nc = 2.0 * 0.30 / 1.80 * math.log10(200.0 / 100.0)
record_settlement_test(
    "TEST 1 - NC",
    "Normally consolidated soil; virgin compression using Cc.",
    result_nc["layer_results"][0]["settlement_m"],
    expected_nc
)

# TEST 2 — OC recompression only
result_oc_recompression = calculate_layered_settlement(
    layer_thickness=[2.0],
    ocr=[2.0],
    cc=[0.30],
    cr=[0.05],
    e0=[0.80],
    sigma_v0_eff=[100.0],
    delta_sigma_eff=[50.0]
)
expected_oc_recompression = 2.0 * 0.05 / 1.80 * math.log10(150.0 / 100.0)
record_settlement_test(
    "TEST 2 - OC Recompression",
    "Overconsolidated soil; final stress remains below preconsolidation pressure.",
    result_oc_recompression["layer_results"][0]["settlement_m"],
    expected_oc_recompression
)

# TEST 3 — OC recompression plus virgin compression
result_oc_combined = calculate_layered_settlement(
    layer_thickness=[2.0],
    ocr=[2.0],
    cc=[0.30],
    cr=[0.05],
    e0=[0.80],
    sigma_v0_eff=[100.0],
    delta_sigma_eff=[150.0]
)
expected_oc_combined = (
    2.0 / 1.80
    * (
        0.05 * math.log10(200.0 / 100.0)
        + 0.30 * math.log10(250.0 / 200.0)
    )
)
record_settlement_test(
    "TEST 3 - OC Combined",
    "Overconsolidated soil; final stress exceeds preconsolidation pressure.",
    result_oc_combined["layer_results"][0]["settlement_m"],
    expected_oc_combined
)

# TEST 4 — Reject OCR below 1
try:
    calculate_layered_settlement(
        layer_thickness=[2.0],
        ocr=[0.8],
        cc=[0.30],
        cr=[0.05],
        e0=[0.80],
        sigma_v0_eff=[100.0],
        delta_sigma_eff=[50.0]
    )
    invalid_ocr_rejected = False
except ValueError:
    invalid_ocr_rejected = True

assert invalid_ocr_rejected, "TEST 4 FAILED: OCR < 1 was not rejected."
settlement_test_results.append({
    "Test Case": "TEST 4 - Invalid OCR",
    "Description": "OCR below 1 must be rejected by the current NC/OC model.",
    "Expected (mm)": None,
    "Actual (mm)": None,
    "Difference (mm)": None,
    "Relative Error (%)": None,
    "Status": "PASS"
})

settlement_test_df = pd.DataFrame(settlement_test_results)
display(settlement_test_df.style.format({
    "Expected (mm)": "{:.6f}",
    "Actual (mm)": "{:.6f}",
    "Difference (mm)": "{:.10f}",
    "Relative Error (%)": "{:.8f}"
}, na_rep="-"))

settlement_passed = sum(row["Status"] == "PASS" for row in settlement_test_results)
print(f"\nLayered Settlement Function — Total: {len(settlement_test_results)} | Passed: {settlement_passed} | Failed: {len(settlement_test_results) - settlement_passed}")
assert settlement_passed == len(settlement_test_results)
print("LAYERED SETTLEMENT FUNCTION: ALL TESTS PASSED")


# ============================================================
# M1 - INITIAL STRESS FUNCTION TESTING
# ============================================================

initial_stress_test_results = []

def record_stress_test(name, description, actual, expected):
    difference = abs(actual - expected)
    passed = math.isclose(actual, expected, rel_tol=1e-10, abs_tol=1e-10)

    initial_stress_test_results.append({
        "Test Case": name,
        "Description": description,
        "Expected (kPa)": expected,
        "Actual (kPa)": actual,
        "Difference (kPa)": difference,
        "Status": "PASS" if passed else "FAIL"
    })

    assert passed, f"{name} FAILED: expected {expected}, actual {actual}."


# TEST 1 — Above groundwater table
initial_stress_above = calculate_initial_stress(
    layer_thickness=[2.0],
    gamma_moist=[17.0],
    gamma_sat=[19.0],
    groundwater_depth=2.0
)
record_stress_test(
    "TEST 1 - Above GWT",
    "Layer-center effective stress above groundwater table.",
    initial_stress_above["layer_results"][0]["sigma_v0_eff_kpa"],
    17.0
)

# TEST 2 — Below groundwater table
initial_stress_below = calculate_initial_stress(
    layer_thickness=[2.0, 3.0],
    gamma_moist=[17.0, 18.0],
    gamma_sat=[19.0, 19.5],
    groundwater_depth=2.0
)
# At z = 3.5 m:
# sigma_v0 = 2.0(17.0) + 1.5(19.5) = 63.25 kPa
# u0 = 9.81(3.5 - 2.0) = 14.715 kPa
# sigma'_v0 = 63.25 - 14.715 = 48.535 kPa
record_stress_test(
    "TEST 2 - Below GWT",
    "Layer-center effective stress below groundwater table.",
    initial_stress_below["layer_results"][1]["sigma_v0_eff_kpa"],
    48.535
)

initial_stress_test_df = pd.DataFrame(initial_stress_test_results)
display(initial_stress_test_df.style.format({
    "Expected (kPa)": "{:.4f}",
    "Actual (kPa)": "{:.4f}",
    "Difference (kPa)": "{:.10f}"
}))

initial_stress_passed = sum(row["Status"] == "PASS" for row in initial_stress_test_results)
print(f"\nInitial Stress Function — Total: {len(initial_stress_test_results)} | Passed: {initial_stress_passed} | Failed: {len(initial_stress_test_results) - initial_stress_passed}")
assert initial_stress_passed == len(initial_stress_test_results)
print("INITIAL STRESS FUNCTION: ALL TESTS PASSED")


# ============================================================
# M1 - VACUUM STRESS-INCREMENT FUNCTION TESTING
# ============================================================

vacuum_test_results = []

def record_vacuum_test(name, description, actual, expected):
    difference = abs(actual - expected)
    passed = math.isclose(actual, expected, rel_tol=1e-10, abs_tol=1e-10)

    vacuum_test_results.append({
        "Test Case": name,
        "Description": description,
        "Expected (kPa)": expected,
        "Actual (kPa)": actual,
        "Difference (kPa)": difference,
        "Status": "PASS" if passed else "FAIL"
    })

    assert passed, f"{name} FAILED: expected {expected}, actual {actual}."


# TEST 1 — Vacuum only; total stress unchanged
vacuum_only = calculate_stress_increment(
    initial_pore_pressure=[20.0, 40.0],
    final_pore_pressure=[-10.0, 10.0]
)
record_vacuum_test(
    "TEST 1 - Vacuum Only",
    "Effective stress increment equals the reduction in pore-water pressure.",
    vacuum_only["layer_results"][0]["delta_sigma_eff_kpa"],
    30.0
)
record_vacuum_test(
    "TEST 2 - Vacuum Only, Layer 2",
    "Second layer receives a different pore-pressure reduction.",
    vacuum_only["layer_results"][1]["delta_sigma_eff_kpa"],
    30.0
)

# TEST 3 — Vacuum plus independently calculated fill stress
vacuum_plus_fill = calculate_stress_increment(
    initial_pore_pressure=[20.0],
    final_pore_pressure=[-10.0],
    delta_sigma_total=[15.0]
)
record_vacuum_test(
    "TEST 3 - Vacuum + Fill",
    "Effective increment includes 15 kPa total-stress increase plus 30 kPa vacuum contribution.",
    vacuum_plus_fill["layer_results"][0]["delta_sigma_eff_kpa"],
    45.0
)

# TEST 4 — Reject pore-pressure increase for the vacuum-loading model
try:
    calculate_stress_increment(
        initial_pore_pressure=[20.0],
        final_pore_pressure=[25.0]
    )
    pore_pressure_increase_rejected = False
except ValueError:
    pore_pressure_increase_rejected = True

assert pore_pressure_increase_rejected, (
    "TEST 4 FAILED: pore-pressure increase was not rejected."
)
vacuum_test_results.append({
    "Test Case": "TEST 4 - Invalid Pore Pressure Change",
    "Description": "A pore-pressure increase is not accepted as vacuum-induced suction.",
    "Expected (kPa)": None,
    "Actual (kPa)": None,
    "Difference (kPa)": None,
    "Status": "PASS"
})

vacuum_test_df = pd.DataFrame(vacuum_test_results)
display(vacuum_test_df.style.format({
    "Expected (kPa)": "{:.4f}",
    "Actual (kPa)": "{:.4f}",
    "Difference (kPa)": "{:.10f}"
}, na_rep="-"))

vacuum_passed = sum(row["Status"] == "PASS" for row in vacuum_test_results)
print(f"\nVacuum Stress-Increment Function — Total: {len(vacuum_test_results)} | Passed: {vacuum_passed} | Failed: {len(vacuum_test_results) - vacuum_passed}")
assert vacuum_passed == len(vacuum_test_results)
print("VACUUM STRESS-INCREMENT FUNCTION: ALL TESTS PASSED")


# ============================================================
# M1 - INTEGRATION TESTING
# ============================================================

# The integrated output must be consistent with the independently
# calculated initial stress, vacuum stress increment, and settlement.

integrated_test = calculate_m1_vacuum_settlement(
    layer_thickness=[2.0],
    gamma_moist=[17.0],
    gamma_sat=[19.0],
    groundwater_depth=0.0,
    ocr=[1.0],
    cc=[0.30],
    cr=[0.05],
    e0=[0.80],
    final_pore_pressure=[-10.0]
)

expected_initial_effective_stress = 19.0 * 1.0 - 9.81 * 1.0
expected_vacuum_increment = 9.81 * 1.0 - (-10.0)
expected_final_effective_stress = expected_initial_effective_stress + expected_vacuum_increment
expected_integrated_settlement = (
    2.0 * 0.30 / 1.80
    * math.log10(expected_final_effective_stress / expected_initial_effective_stress)
)

assert math.isclose(
    integrated_test["layer_results"][0]["sigma_v0_eff_kpa"],
    expected_initial_effective_stress,
    rel_tol=1e-10,
    abs_tol=1e-10
)
assert math.isclose(
    integrated_test["layer_results"][0]["delta_sigma_eff_kpa"],
    expected_vacuum_increment,
    rel_tol=1e-10,
    abs_tol=1e-10
)
assert math.isclose(
    integrated_test["layer_results"][0]["sigma_vf_eff_kpa"],
    expected_final_effective_stress,
    rel_tol=1e-10,
    abs_tol=1e-10
)
assert math.isclose(
    integrated_test["total_settlement_mm"],
    expected_integrated_settlement * 1000.0,
    rel_tol=1e-10,
    abs_tol=1e-10
)

integration_report = pd.DataFrame([{
    "Check": "Integrated M1 calculation",
    "Initial effective stress (kPa)": integrated_test["layer_results"][0]["sigma_v0_eff_kpa"],
    "Effective stress increment (kPa)": integrated_test["layer_results"][0]["delta_sigma_eff_kpa"],
    "Final effective stress (kPa)": integrated_test["layer_results"][0]["sigma_vf_eff_kpa"],
    "Settlement (mm)": integrated_test["total_settlement_mm"],
    "Status": "PASS"
}])

display(integration_report.style.format({
    "Initial effective stress (kPa)": "{:.4f}",
    "Effective stress increment (kPa)": "{:.4f}",
    "Final effective stress (kPa)": "{:.4f}",
    "Settlement (mm)": "{:.6f}"
}))

print("M1 INTEGRATION TEST: ALL CHECKS PASSED")
print("\nM1 FUNCTION TESTING AND INTEGRATION TESTING COMPLETED")
```
